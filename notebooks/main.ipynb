```python
import jax.numpy as jnp
import matplotlib.pyplot as plt
import jax
from jax.scipy.special import bessel_jn
from fft import *

jax.config.update("jax_enable_x64", True)
```

```python
jax.devices()
```

```python
W = 40
H = 40
X, Y, R = create_axes(W, H, 1)
```

```python
# Airy disk
E_airy = 2 * bessel_jn(jnp.abs(2 * R), v=1)[1, :] / jnp.abs(2 * R)
```

```python
plt.imshow(E_airy, extent=[-W / 2, W / 2, -H / 2, H / 2])
# plt.ylim(-0.5,1.2)
# plt.axhline(0,color='black',linestyle='--')
```

```python
F_airy = fft2_shifted(E_airy)
```

```python
plt.imshow(jnp.abs(F_airy))
```
