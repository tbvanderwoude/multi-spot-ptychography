```python
import sympy as sp
```

```python
# See https://docs.sympy.org/latest/guides/assumptions.html for more assumptions
epsilon = sp.Symbol("epsilon", positive=True)
expr_1 = sp.acos(1 - epsilon)
sp.series(expr_1, epsilon, 0, 5)
```

```python
expr_2 = sp.sqrt(1 - (1 - epsilon) ** 2)
sp.series(expr_2, epsilon, 0, 5)
```

So for this standard OTF, indeed $\hat{g}(\vec{q}) \propto (2NA/\lambda - |\vec{q}|)$.

```python
g_expr = 2 / sp.pi * (expr_1 - (1 - epsilon) * expr_2)
s = sp.series(g_expr, epsilon, 0, 5)
print(sp.latex(s))
s
```
