```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

x = np.linspace(-2, 6, 1000)
theta_hat_mean = 2
theta_hat_var = 1
theta_hat_CRB = 0.5
theta_true = 2
for theta_hat_mean in [0, 2]:
    for theta_hat_var in [1, 0.5]:
        # if theta_hat_CRB != theta_hat_var:
        #     plt.plot(x, norm.pdf(x,theta_hat_mean,theta_hat_var),label=r"$\hat{\theta}$ PDF")
        #     plt.plot(x, norm.pdf(x,theta_hat_mean,theta_hat_CRB),linestyle='--',label=r"Efficient $\hat{\theta}$ PDF")
        # else:
        #     plt.plot(x, norm.pdf(x,theta_hat_mean,theta_hat_var),label=r"Efficient $\hat{\theta}$ PDF")

        plt.xlabel(r"$\hat{\theta}$")
        plt.ylabel(r"$p(\hat{\theta})$")
        if theta_true != theta_hat_mean:
            plt.plot(
                x,
                norm.pdf(x, theta_hat_mean, theta_hat_var),
                label=r"$\hat{\theta}$ PDF",
            )
            plt.axvline(
                theta_true,
                linestyle="--",
                color="r",
                label=rf"True $\theta={theta_true}$ ",
            )
            plt.axvline(
                theta_hat_mean,
                linestyle="dotted",
                color="black",
                label=r"$\mathbb{E}[\hat{\theta}]=" + rf"{theta_hat_mean}$",
            )
        else:
            if theta_hat_CRB != theta_hat_var:
                plt.plot(
                    x,
                    norm.pdf(x, theta_hat_mean, theta_hat_var),
                    label=r"$\hat{\theta}$ PDF",
                )
                plt.plot(
                    x,
                    norm.pdf(x, theta_hat_mean, theta_hat_CRB),
                    linestyle="--",
                    label=r"Efficient $\hat{\theta}$ PDF",
                )
            else:
                plt.plot(
                    x,
                    norm.pdf(x, theta_hat_mean, theta_hat_var),
                    label=r"Efficient $\hat{\theta}$ PDF",
                )
            plt.axvline(
                theta_true,
                linestyle="--",
                color="r",
                label=r"$\mathbb{E}[\hat{\theta}]=\theta=" + f"{theta_true}$ ",
            )
        plt.legend()
        # plt.grid(True)
        plt.ylim(0.0, 1.0)
        plt.show()
```

```python
# AI-generated variant of the above

plt.rcParams.update(
    {
        "font.size": 16,
        "axes.labelsize": 18,
        "legend.fontsize": 14,
        "lines.linewidth": 2.5,
        "mathtext.fontset": "cm",
        "savefig.dpi": 300,
    }
)

x = np.linspace(-2, 6, 1000)
theta_true, crb = 2, 0.25
pdf = lambda m, v: norm.pdf(x, m, np.sqrt(v))  # v = variance

for mean in [0, 2]:
    for var in [1, np.sqrt(1 / 2)]:
        fig, ax = plt.subplots(figsize=(9, 4.5), layout="constrained")
        if mean != theta_true:
            ax.plot(x, pdf(mean, var), label=r"$\hat\theta$ PDF")
            ax.axvline(theta_true, ls="--", c="r", label=rf"True $\theta={theta_true}$")
            ax.axvline(mean, ls=":", c="k", label=rf"$\mathbb{{E}}[\hat\theta]={mean}$")
        else:
            ax.plot(
                x,
                pdf(mean, var),
                label=(
                    r"$\hat\theta$ PDF" if var != crb else r"Efficient $\hat\theta$ PDF"
                ),
            )
            if var != crb:
                ax.plot(x, pdf(mean, crb), ls="--", label=r"Efficient $\hat\theta$ PDF")
            ax.axvline(
                theta_true,
                ls="--",
                c="r",
                label=rf"$\mathbb{{E}}[\hat\theta]=\theta={theta_true}$",
            )
        ax.set(
            xlabel=r"$\hat\theta$",
            ylabel=r"$p(\hat\theta)$",
            xlim=(x[0], x[-1]),
            ylim=(0, 1),
        )
        ax.legend(loc="upper right")
        fig.savefig(f"pdf_mean{mean}_var{var}.png", transparent=True)
        plt.close(fig)
```

```python
cases = [
    (0, 1, True),
    (0, 0.25, True),
    (2, 1, False),
    (2, 1, True),
    (2, 0.25, True),
]

for mean, var, eff in cases:
    fig, ax = plt.subplots(figsize=(9, 4.5), layout="constrained")
    if mean != theta_true:
        ax.plot(x, pdf(mean, var), label=r"$\hat\theta$ PDF")
        ax.axvline(theta_true, ls="--", c="r", label=rf"True $\theta={theta_true}$")
        ax.axvline(mean, ls=":", c="k", label=rf"$\mathbb{{E}}[\hat\theta]={mean}$")
    else:
        ax.plot(
            x,
            pdf(mean, var),
            label=r"$\hat\theta$ PDF" if var != crb else r"Efficient $\hat\theta$ PDF",
        )
        if eff and var != crb:
            ax.plot(x, pdf(mean, crb), ls="--", label=r"Efficient $\hat\theta$ PDF")
        ax.axvline(
            theta_true,
            ls="--",
            c="r",
            label=rf"$\mathbb{{E}}[\hat\theta]=\theta={theta_true}$",
        )
    ax.set(
        xlabel=r"$\hat\theta$",
        ylabel=r"$p(\hat\theta)$",
        xlim=(x[0], x[-1]),
        ylim=(0, 1),
    )
    ax.legend(loc="upper right")
    fig.savefig(
        f"pdf_mean{mean}_var{var}{'' if eff else '_noeff'}.png", transparent=True
    )
    plt.close(fig)
```
